File Handling

### Writing to a file

First, let's write some text to a file. We'll open the file in write mode (`'w'`), which creates the file if it doesn't exist or overwrites it if it does.

```python
file_name = 'my_example_file.txt'
content_to_write = "Hello, this is a test line.\nThis is another line of text.\nAnd a final line."

with open(file_name, 'w') as file:
    file.write(content_to_write)

print(f"Content successfully written to '{file_name}'")
```

### Reading from a file

Now, let's read the content back from the file. We'll open the file in read mode (`'r'`).

```python
file_name = 'my_example_file.txt'

with open(file_name, 'r') as file:
    read_content = file.read()

print(f"Content read from '{file_name}':\n")
print(read_content)
```

### Appending to a file

If you want to add content to an existing file without overwriting it, you can open it in append mode (`'a'`).

```python
file_name = 'my_example_file.txt'
additional_content = "\nThis line was appended."

with open(file_name, 'a') as file:
    file.write(additional_content)

print(f"Content successfully appended to '{file_name}'")

# Read again to see the appended content
with open(file_name, 'r') as file:
    read_content_after_append = file.read()

print(f"\nContent of '{file_name}' after appending:\n")
print(read_content_after_append)
```

4. Write a program to handle the error when a file does not exist using exception handling.

```python
file_name = 'file2.txt'
try:
    with open(file_name, 'r') as file:
        content = file.read()
        print(f"Content of '{file_name}':\n{content}")
except FileNotFoundError:
    print(f"Error: The file '{file_name}' was not found. Please check the file name and path.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

5. Explain the difference between write ("w") mode and append ("a") mode with a small code example.

```python
file_name = 'mode_comparison.txt'
print("--- Demonstrating 'w' mode (overwrite) ---")
with open(file_name, 'w') as file:
    file.write("First line written in 'w' mode.\n")
print(f"Content after initial 'w' write:\n")
with open(file_name, 'r') as file:
    print(file.read())
with open(file_name, 'w') as file:
    file.write("Second line, overwriting previous content in 'w' mode.\n")
print(f"\nContent after second 'w' write (overwritten):\n")
with open(file_name, 'r') as file:
    print(file.read())
print("--- Demonstrating 'a' mode (append) ---")
with open(file_name, 'a') as file:
    file.write("Third line, appended in 'a' mode.\n")
print(f"\nContent after first 'a' append:\n")
with open(file_name, 'r') as file:
    print(file.read())
with open(file_name, 'a') as file:
    file.write("Fourth line, also appended in 'a' mode.\n")
print(f"\nContent after second 'a' append:\n")
with open(file_name, 'r') as file:
    print(file.read())
```

### Object-Oriented Programming (OOP) in Python

OOP is a way of structuring programs by bundling related properties and behaviors into individual objects. Key concepts include:

*   **Class**: A blueprint for creating objects. It defines a set of attributes (data) and methods (functions) that the objects created from the class will have.
*   **Object**: An instance of a class. When a class is defined, no memory is allocated until an object of that class is created.
*   **Encapsulation**: The bundling of data and methods that operate on the data within a single unit (e.g., a class), and restricting direct access to some of the object's components.
*   **Inheritance**: A mechanism that allows a new class (subclass) to inherit properties and behaviors from an existing class (superclass), promoting code reuse.
*   **Polymorphism**: The ability of an object to take on many forms. It allows methods to do different things depending on the object it is called on.
*   **Abstraction**: Hiding the complex implementation details and showing only the necessary and relevant parts to the user.

```python
class Dog:
    species = "Canis familiaris"
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def bark(self):
        return f"{self.name} says Woof!"
    def description(self):
        return f"{self.name} is {self.age} years old."
my_dog = Dog("Buddy", 3)
your_dog = Dog("Lucy", 5)
print(f"My dog's name is {my_dog.name} and he is {my_dog.age} years old.")
print(f"Your dog's name is {your_dog.name} and she is {your_dog.age} years old.")
print(my_dog.bark())
print(your_dog.description())
print(f"Buddy is a {my_dog.species}")
```

```python
print("### Understanding Exception Handling")
print("Exception handling allows you to manage errors that occur during program execution. It uses `try`, `except`, `else`, and `finally` blocks.")

def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Both inputs must be numbers!")
        return None
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return None
    else:
        print("Division successful!")
        return result
    finally:
        print("Execution of divide_numbers function complete.")
print("\n--- Test Cases ---")
print("\nCalling divide_numbers(10, 2):")
output1 = divide_numbers(10, 2)
if output1 is not None:
    print(f"Result: {output1}")
print("\nCalling divide_numbers(10, 0):")
output2 = divide_numbers(10, 0)
if output2 is not None:
    print(f"Result: {output2}")
print("\nCalling divide_numbers(10, 'a'):")
output3 = divide_numbers(10, 'a')
if output3 is not None:
    print(f"Result: {output3}")
print("\nCalling divide_numbers(15, 3):")
output4 = divide_numbers(15, 3)
if output4 is not None:
    print(f"Result: {output4}")
```
