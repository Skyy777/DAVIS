You are given a NumPy array of student marks. Write a program to extract marks greater than the class average.

```python
import numpy as np
marks = np.array([75, 82, 90, 65, 88, 70, 95, 78])
class_average = np.mean(marks)
print(f"Original marks: {marks}")
print(f"Class average: {class_average:.2f}")
marks_above_average = marks[marks > class_average]
print(f"Marks greater than the class average: {marks_above_average}")
```

Given a NumPy array of sensor readings, write a program to detect and extract all abnormal readings that are more than 2 times the mean value.

```python
sensor_readings = np.array([10, 12, 11, 15, 100, 13, 14, 12, 200, 16])
print(f"Original sensor readings: {sensor_readings}")
mean_value = np.mean(sensor_readings)
print(f"Mean value: {mean_value:.2f}")
abnormal_threshold = 2 * mean_value
print(f"Abnormal threshold (2 * mean): {abnormal_threshold:.2f}")
abnormal_readings = sensor_readings[sensor_readings > abnormal_threshold]
print(f"Abnormal readings (more than 2 times the mean): {abnormal_readings}")
```

A NumPy array contains salary data of employees. Write a program to apply tax rules:


Salary above 50,000 → deduct 10 percent


Salary above 1,00,000 → deduct 20 percent

A NumPy array contains ratings given by users. Write a program to normalize the ratings between 0 and 1

```python
ratings = np.array([4, 3, 5, 2, 4, 3, 5])
print(f"Original ratings: {ratings}")
min_rating = np.min(ratings)
max_rating = np.max(ratings)
if max_rating == min_rating:
    normalized_ratings = np.zeros_like(ratings)
else:
    normalized_ratings = (ratings - min_rating) / (max_rating - min_rating)
print(f"Normalized ratings between 0 and 1: {normalized_ratings}")
```
