You are given a NumPy array of student marks. Write a program to extract marks greater than the class average.

```python
import numpy as np
marks = np.array([75, 82, 90, 65, 88, 70, 95, 78])
class_average = np.mean(marks)
print(f"Original marks: {marks}")
print(f"Class average: {class_average:.2f}")
marks_above_average = marks[marks > class_average]
print(f"Marks greater than the class average: {marks_above_average}")
```

```python
sales_data = np.array([45000, 52000, 48000, 60000, 39000, 55000, 51000])
print(f"Original sales data: {sales_data}")
high_sales_days = sales_data[sales_data > 50000]
print(f"Days with sales higher than 50,000: {high_sales_days}")
```

Given a NumPy array of temperatures in Celsius, write a program to convert all values to Fahrenheit using vectorized operations.

```python
celsius_temps = np.array([0, 10, 20, 25, 30, 37, 100])
print(f"Original Celsius temperatures: {celsius_temps}")
fahrenheit_temps = celsius_temps * 9/5 + 32
print(f"Converted Fahrenheit temperatures: {fahrenheit_temps}")
```

Given a NumPy array of product prices, write a program to apply a 10 percent discount on prices greater than 1,000

```python
product_prices = np.array([500, 1200, 800, 1500, 950, 2000, 1100])
print(f"Original product prices: {product_prices}")
high_price_mask = product_prices > 1000
product_prices[high_price_mask] = product_prices[high_price_mask] * 0.90
print(f"Product prices after 10% discount on prices > 1,000: {product_prices}")
```

Given a 2D NumPy array representing marks of students in multiple subjects, write a program to calculate student-wise average marks

```python
student_marks = np.array([
    [85, 90, 78, 92],
    [70, 65, 80, 75],
    [95, 88, 91, 87],
    [60, 72, 68, 70]
])
print(f"Original student marks:\n{student_marks}")
student_average_marks = np.mean(student_marks, axis=1)
print(f"\nStudent-wise average marks: {student_average_marks}")
```

A NumPy array contains monthly expenses. Write a program to identify expenses that are above the median value

```python
monthl_expenses = np.array([1500, 2000, 1800, 2200,2500])
print(f"Original monthly expenses: {monthl_expenses}")
median_expense = np.median(monthl_expenses)
print(f"Median expense: {median_expense}")
above_median_expenses = monthl_expenses[monthl_expenses > median_expense]
print(f"Expenses above the median value: {above_median_expenses}")
```

A 2D NumPy array represents sales data of products across different regions. Write a program to find the region with the highest total sales.

```python
sales_data = np.array([
    [100, 150, 120, 200],
    [130, 110, 180, 160],
    [210, 90, 140, 170],
    [180, 220, 100, 190]
])
print(f"Original sales data (regions x products):\n{sales_data}")
total_sales_per_region = np.sum(sales_data, axis=1)
print(f"\nTotal sales per region: {total_sales_per_region}")
highest_sales_region_index = np.argmax(total_sales_per_region)
highest_sales_region_number = highest_sales_region_index + 1
print(f"Region with the highest total sales: Region {highest_sales_region_number} (Total Sales: {total_sales_per_region[highest_sales_region_index]})")
```
