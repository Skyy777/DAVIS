 B. Line Chart Practice Questions


Plot a line chart showing monthly electricity consumption for 12 months

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = np.arange(1, 13)
electricity_consumption = np.random.normal(200, 500, size=12)
plt.plot(months, electricity_consumption,color='blue')
plt.title('Monthly Electricity Consumption (12 Months)')
plt.xlabel('Month')
plt.ylabel('Electricity Consumption (kWh)')
plt.xticks(months)
plt.show()
```

Create a line chart to show the growth of website users over 8 months.

```python
users = np.random.normal(50, 200, size=8)
months = np.arange(1, 9)
plt.plot(months, users, color='green')
plt.title('Website Users Growth Over 8 Months')
plt.xlabel('Month')
plt.ylabel('Number of Users')
plt.xticks(months)
plt.show()
```

Draw a line chart for daily stock prices and identify whether the trend is upward or downward.

```python
stork_prices = np.random.normal(100, 20, size=30)
days = np.arange(1, 31)
plt.plot(days, stork_prices, color='red')
plt.title('Daily Stock Prices')
plt.xlabel('Day')
plt.ylabel('Stock Price ($)')
plt.xticks(days)
plt.show()
```

Plot a line chart to compare temperature changes of two cities over 7 days.

```python
temerature_city1 = np.random.normal(25, 5, size=7)
temerature_city2 = np.random.normal(20, 3, size=7)
days = np.arange(1, 8)
plt.plot(days, temerature_city1, label='City 1', color='blue')
plt.plot(days, temerature_city2, label='City 2', color='green')
plt.title('Temperature Changes of Two Cities Over 7 Days')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.xticks(days)
plt.legend()
plt.show()
```

Create a line chart showing progress of a student’s marks across 5 exams.

```python
student_marks = (20,40,50,60,40)
exams = ['Exam 1', 'Exam 2', 'Exam 3', 'Exam 4', 'Exam 5']
plt.plot(exams, student_marks, color='orange')
plt.title('Student Marks Progress Across Exams')
plt.xlabel('Exam')
plt.ylabel('Marks')
plt.show()
```

 C. Bar Chart Practice Questions


Create a bar chart showing marks scored by 5 students in an exam

```python
marks=(20,40,50,60,40)
students = ['student a', 'student b', 'student c', 'student d', 'student e']
plt.bar(students, marks, color='purple')
plt.title('Marks Scored by 5 Students')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.show()
```

Plot a bar chart comparing sales of three different brands

```python
sales = (200, 300, 150)
brands = ['Brand A', 'Brand B', 'Brand C']
plt.bar(brands, sales, color='yellow')
plt.title('Sales of Three Different Brands')
plt.xlabel('Brand')
plt.ylabel('Sales')
plt.show()
```

Draw a bar chart to represent number of employees in different departments.

```python
employees = (200, 300, 150)
departments = ['Department A', 'Department B', 'Department C']
plt.bar(departments, employees, color='red')
plt.title('Number of Employees in Different Departments')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```

Create a bar chart to compare votes received by different political parties.

```python
votes = (200, 300, 150)
parties = ['Party A', 'Party B', 'Party C']
plt.bar(parties, votes, color='orange')
plt.title('Votes Received by Different Political Parties')
plt.xlabel('Political Party')
plt.ylabel('Votes')
plt.show()
```

Plot a bar chart showing product-wise profit in a company.

```python
profit = (200, 300, 150,250,500,300)
product = ['Product A', 'Product B', 'Product C','Product D','Product E','Product F']
plt.bar(product, profit, color='red')
plt.title('Product-wise Profit in a Company')
plt.xlabel('Product')
plt.ylabel('Profit')
plt.show()
```

 D. Histogram Practice Questions

Create a histogram to show the distribution of marks of students in a class.


```python
marks = (20,40,50,60,40,30,35)
plt.hist(marks, bins=5, color='black')
plt.title('Distribution of Marks in a Class')
plt.xlabel('Marks')
plt.ylabel('Frequency')
```

Plot a histogram to analyze age distribution of people in a survey.

```python
age=np.random.normal(20, 5, size=100)
plt.hist(age, bins=10, color='green')
plt.title('Age Distribution in a Survey')
plt.xlabel('Age')
plt.ylabel('Frequency')
```

Draw a histogram for salary distribution of employees in an organization

```python
salary = np.random.normal(50000, 10000, size=100)
plt.hist(salary, bins=10, color='blue')
plt.title('Salary Distribution of Employees')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

Create a histogram to analyze website response times.

```python
response_times = np.random.normal(10, 2, size=100)
plt.hist(response_times, bins=10, color='orange')
plt.title('Website Response Times')
plt.xlabel('Response Time (seconds)')
plt.ylabel('Frequency')
plt.show()
```

Plot a histogram to understand how exam scores are spread across different ranges

```python
scores = np.random.normal(70, 10, 100)
plt.figure(figsize=(10, 6))
plt.hist(scores, bins=10, color='purple')
plt.title('Exam Scores Spread')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
```
