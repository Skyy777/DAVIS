Write a function in python to read the content from a text file "ABC.txt" line by line
and display the same on screen.

```python
def read_and_display_file_line_by_line(filename="ABC.txt"):
    try:
        with open(filename, 'r') as file:
            for line_number, line_content in enumerate(file, 1):
                print(f"Line {line_number}: {line_content.strip()}")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
with open('ABC.txt', 'w') as f:
    f.write("This is the first line.\n")
    f.write("This is the second line.\n")
    f.write("And a third line.")
read_and_display_file_line_by_line("ABC.txt")
```

Write a function in Python to count and display the total number of words in a text
file “ABC.txt”

```python
def count_words_in_file(filename="ABC.txt"):
    word_count = 0
    try:
        with open(filename, 'r') as file:
            for line in file:
                words = line.split()
                word_count += len(words)
        print(f"Total number of words in '{filename}': {word_count}")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
count_words_in_file("ABC.txt")
```

Write a function in Python to count uppercase character in a text file “ABC.txt”

```python
def count_uppercase_chars_in_file(filename="ABC.txt"):
    uppercase_count = 0
    try:
        with open(filename, 'r') as file:
            for line in file:
                for char in line:
                    if char.isupper():
                        uppercase_count += 1
        print(f"Total number of uppercase characters in '{filename}': {uppercase_count}")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
count_uppercase_chars_in_file("ABC.txt")
```

Write a function display_words() in python to read lines from a text file "story.txt",
and display those words, which are less than 4 characters.

```python
def display_words(filename="story.txt"):
    try:
        with open(filename, 'r') as file:
            for line in file:
                words = line.replace('.', '').replace(',', '').split()
                for word in words:
                    if len(word) < 4:
                        print(word)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")
with open('story.txt', 'w') as f:
    f.write("Once upon a time, there was a little bird.\n")
    f.write("The bird flew high in the sky. And it was free.\n")
    f.write("It sang a sweet song.")
display_words("story.txt")
```

Write a Python program to handle a ZeroDivisionError exception when dividing a
number by zero.

```python
def divide_numbers(numerator, denominator):
    try:
        result = numerator / denominator
        print(f"The result of {numerator} / {denominator} is: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
    except TypeError:
        print("Error: Please ensure both inputs are numbers.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
divide_numbers(10, 2)
divide_numbers(5, 0)
divide_numbers(10, -5)
divide_numbers(8, 'a')
```

Write a Python program that prompts the user to input an integer and raises a
ValueError exception if the input is not a valid integer.

```python
def get_integer_input():
    while True:
        try:
            user_input = input("Please enter an integer: ")
            value = int(user_input)
            print(f"You entered a valid integer: {value}")
            break
        except ValueError:
            print("Error: Invalid input. Please enter a whole number.")
        except Exception as e:
            print(f"An unexpected error occurred: {e}")
get_integer_input()
```

Write a Python program that opens a file and handles a FileNotFoundError exception
if the file does not exist.

```python
def open_file_safely(filename):
    try:
        with open(filename, 'r') as file:
            print(f"Successfully opened '{filename}'. Content preview:")
            for _ in range(3):
                line = file.readline()
                if not line:
                    break
                print(line.strip())
    except FileNotFoundError:
        print(f"Error: The file '{filename}' not found.")
    except Exception as e:
        print(f" unexpected error  '{filename}': {e}")
print("\nTesting")
open_file_safely('ABC.txt')
print("\n Testing ")
open_file_safely('non_existent_file.txt')
```

Write a Python program that prompts the user to input two numbers and raises a
TypeError exception if the inputs are not numerical

```python
def get_and_validate_numbers():
    while True:
        try:
            input1 = input("Enter the first number: ")
            input2 = input("Enter the second number: ")
            try:
                num1 = float(input1)
            except ValueError:
                raise TypeError(f"Input '{input1}' is not a numerical value.")
            try:
                num2 = float(input2)
            except ValueError:
                raise TypeError(f"Input '{input2}' is not a numerical value.")
            print(f"You entered valid numbers: {num1} and {num2}")
            print(f"Their sum is: {num1 + num2}")
            break
        except TypeError as e:
            print(f"Error: {e} Please try again.")
        except Exception as e:
            print(f"An unexpected error occurred: {e}")
get_and_validate_numbers()
```
