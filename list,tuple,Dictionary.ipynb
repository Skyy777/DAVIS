A **list** is a mutable, ordered sequence of items. This means you can change its contents (add, remove, or modify items) after it's created, and the order of elements is preserved. Lists are defined using square brackets `[]`.

```python
# Creating a list
my_list = [1, 2, 'hello', True]
print(f"Initial list: {my_list}")

# Accessing elements (zero-indexed)
print(f"First element: {my_list[0]}")

# Modifying an element
my_list[2] = 'world'
print(f"List after modification: {my_list}")

# Adding an element
my_list.append(3.14)
print(f"List after appending: {my_list}")

# Slicing a list
print(f"Slice of the list (elements from index 1 to 3): {my_list[1:4]}")
```

A **tuple** is an immutable, ordered sequence of items. This means once a tuple is created, you cannot change its contents (add, remove, or modify items). Tuples are defined using parentheses `()`.

```python
# Creating a tuple
my_tuple = (10, 20, 'python', False)
print(f"Initial tuple: {my_tuple}")

# Accessing elements (zero-indexed)
print(f"Third element: {my_tuple[2]}")

# Tuples are immutable, so this would raise an error:
# my_tuple[0] = 5
# print(my_tuple)

# Slicing a tuple
print(f"Slice of the tuple (elements from index 0 to 2): {my_tuple[0:3]}")

# You can concatenate tuples to create a new tuple
another_tuple = (30,)
combined_tuple = my_tuple + another_tuple
print(f"Combined tuple: {combined_tuple}")
```

### Key Differences:

*   **Mutability:** Lists are mutable (changeable), while tuples are immutable (unchangeable).
*   **Syntax:** Lists use square brackets `[]`, tuples use parentheses `()`.
*   **Use Cases:**
    *   **Lists** are often used for collections of items that might change during the program's execution (e.g., a collection of user inputs, a dynamic shopping cart).
    *   **Tuples** are often used for collections of items that should not change (e.g., coordinates, database records, returning multiple values from a function).

### List within a Tuple

A tuple can contain a list as one of its elements. While the tuple itself remains immutable (you cannot change which list it points to, or add/remove elements from the tuple), the list *inside* the tuple remains mutable. This means you can modify the contents of the list even though it's part of an immutable tuple.

```python
# A tuple containing a list
tuple_with_list = (1, 'apple', [10, 20, 30], True)
print(f"Original tuple with list: {tuple_with_list}")

# Accessing the list within the tuple
inner_list = tuple_with_list[2]
print(f"Accessed inner list: {inner_list}")

# Modifying the list inside the tuple (this is allowed)
inner_list.append(40)
print(f"Modified inner list: {inner_list}")
print(f"Tuple after modifying inner list: {tuple_with_list}")

# Trying to modify the tuple itself (this would raise an error)
# tuple_with_list[0] = 5
```

### Tuple within a List

A list can contain a tuple as one of its elements. Since lists are mutable, you can add, remove, or change elements of the list, including adding or removing tuples. The tuples themselves, once created, remain immutable.

```python
# A list containing a tuple
list_with_tuple = ['a', 100, (1, 2, 3), False]
print(f"Original list with tuple: {list_with_tuple}")

# Accessing the tuple within the list
inner_tuple = list_with_tuple[2]
print(f"Accessed inner tuple: {inner_tuple}")

# Trying to modify the tuple inside the list (this would raise an error)
# inner_tuple[0] = 99

# Modifying the list itself (adding a new tuple)
list_with_tuple.append((4, 5))
print(f"List after adding a new tuple: {list_with_tuple}")

# Modifying the list itself (changing an element to a new tuple)
list_with_tuple[0] = ('new_tuple_value',)
print(f"List after changing an element to a new tuple: {list_with_tuple}")
```

A **dictionary** is an unordered collection of items. Each item stored in a dictionary is a **key-value pair**. This means you access values using unique keys, rather than numerical indices (like in lists and tuples). Dictionaries are mutable, meaning you can add, remove, and modify key-value pairs after creation. Dictionaries are defined using curly braces `{}`.

```python
# Creating a dictionary
my_dict = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York'
}
print(f"Initial dictionary: {my_dict}")

# Accessing values using keys
print(f"Alice's age: {my_dict['age']}")

# Modifying a value
my_dict['age'] = 31
print(f"Dictionary after modifying age: {my_dict}")

# Adding a new key-value pair
my_dict['occupation'] = 'Engineer'
print(f"Dictionary after adding occupation: {my_dict}")

# Removing a key-value pair
del my_dict['city']
print(f"Dictionary after deleting city: {my_dict}")

# Getting all keys
print(f"All keys: {my_dict.keys()}")

# Getting all values
print(f"All values: {my_dict.values()}")

# Getting all items (key-value pairs)
print(f"All items: {my_dict.items()}")
```

A **set** is an unordered collection of unique items. This means that each element in a set must be unique; duplicate elements are automatically removed. Sets are mutable, so you can add or remove items after creation. Sets are defined using curly braces `{}` or the `set()` constructor.

```python
# Creating a set
my_set = {1, 2, 3, 2, 4}
print(f"Initial set: {my_set}") # Note: duplicates are removed

# Creating a set from a list
list_to_set = [1, 2, 2, 3, 4, 4, 5]
another_set = set(list_to_set)
print(f"Set created from list: {another_set}")

# Adding elements to a set
my_set.add(5)
print(f"Set after adding 5: {my_set}")
my_set.add(1) # Adding an existing element has no effect
print(f"Set after adding 1 (already exists): {my_set}")

# Removing elements from a set
my_set.remove(3) # Raises KeyError if element not found
print(f"Set after removing 3: {my_set}")

# Discarding elements (safer remove)
my_set.discard(10) # Does nothing if element not found
print(f"Set after discarding 10 (not present): {my_set}")

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union (all unique elements from both sets)
print(f"Union of set_a and set_b: {set_a.union(set_b)}")
print(f"Union using | operator: {set_a | set_b}")

# Intersection (common elements)
print(f"Intersection of set_a and set_b: {set_a.intersection(set_b)}")
print(f"Intersection using & operator: {set_a & set_b}")

# Difference (elements in set_a but not in set_b)
print(f"Difference (set_a - set_b): {set_a.difference(set_b)}")
print(f"Difference using - operator: {set_a - set_b}")

# Symmetric Difference (elements in either set, but not in both)
print(f"Symmetric Difference: {set_a.symmetric_difference(set_b)}")
print(f"Symmetric Difference using ^ operator: {set_a ^ set_b}")
```

### Important characteristics of sets:

*   **Unique Elements:** Sets automatically handle uniqueness; duplicate elements are not stored.
*   **Unordered:** Elements in a set do not have a defined order, and you cannot access them using an index.
*   **Mutable:** You can add or remove elements from a set after it's created.
*   **Immutable Elements:** While the set itself is mutable, the elements within a set must be immutable (e.g., numbers, strings, tuples). Lists or dictionaries cannot be direct elements of a set.
*   **Membership Testing:** Sets are highly optimized for checking if an element is present in the set (e.g., `element in my_set`).
*   **Mathematical Operations:** Sets support mathematical operations like union, intersection, difference, and symmetric difference, which are very efficient.

### Important characteristics of dictionaries:

*   **Key-Value Pairs:** Each item is a pair consisting of a key and its associated value.
*   **Unique Keys:** Keys within a dictionary must be unique. If you try to add an item with an existing key, the old value associated with that key will be overwritten.
*   **Mutable:** Dictionaries can be changed after they are created.
*   **Unordered (in Python versions before 3.7):** Historically, dictionaries were unordered. From Python 3.7 onwards, dictionaries maintain the insertion order of items. However, relying on this order for all operations is generally not recommended unless explicitly needed and understood.
*   **Keys are Immutable:** Dictionary keys must be immutable types, such as strings, numbers, or tuples. Mutable types like lists or other dictionaries cannot be used as keys.
*   **Values can be of any type:** Values can be any Python object, including other dictionaries, lists, or custom objects.

```python
#Write a program to find the union of two sets.

set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")

union_set_method = set_a.union(set_b)
print(f"Union using .union() method: {union_set_method}")

union_set_operator = set_a | set_b
print(f"Union using | operator: {union_set_operator}")
```
