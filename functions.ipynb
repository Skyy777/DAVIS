```python
def greet_person(name, age):
    """This function greets a person and tells them their age."""
    print(f"Hello, {name}! You are {age} years old.")

greet_person("Alice", 30)
greet_person("Bob", 25)
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        res = 1
        for i in range(1, n + 1):
            res *= i
        return res
print(f"Factorial of 5 is: {factorial(5)}")
print(f"Factorial of 0 is: {factorial(0)}")
print(f"Factorial of 7 is: {factorial(7)}")
```

```python
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n-1)

number = int(input("Please enter a number: "))
result = fact(number)
print(f"The factorial of {number} is {fact(number)}")
```
