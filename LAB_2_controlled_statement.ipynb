```python
# leap year
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"The year {year} is a leap year.")
else:
    print(f"The year {year} is not a leap year.")
```

```python
# Read the product code and order amount
product_code = int(input("Enter the product code (1 for Battery Based, 2 for Key-based, 3 for Electrical Charging Based): "))
order_amount = float(input("Enter the order amount in Rs.: "))
discount = 0.0
if product_code == 1:
    if order_amount > 1000:
        discount = 0.10 * order_amount
elif product_code == 2:
    if order_amount > 100:
        discount = 0.05 * order_amount
elif product_code == 3:
    if order_amount > 500:
        discount = 0.10 * order_amount
else:
    print("Invalid product code. Please enter 1, 2, or 3.")
    exit()
net_amount = order_amount - discount
print(f"The net amount to be paid is Rs. {net_amount:.2f}")
```

```python
# Read the distance in km
distance = float(input("Enter the distance in km: "))

fare = 0.0
if distance <= 50:
    fare = distance * 8
elif distance <= 100:
    fare = 50 * 8 + (distance - 50) * 10
else:
    fare = 50 * 8 + 50 * 10 + (distance - 100) * 12
print(f"The total fare is Rs. {fare:.2f}")
```

```python
num = int(input("Enter a number to reverse: "))
reversed_num = 0

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num = num // 10

print(f"The reversed number is: {reversed_num}")
```

```python
#Write a python program to check whether a number is palindrome or not
num = int(input("Enter a number to check for palindrome: "))
num_str = str(num)
reversed_str = num_str[::-1]
if num_str == reversed_str:
    print(f"{num} is a palindrome.")
else:
    print(f"{num} is not a palindrome.")
```

```python
# Write a python program finding the factorial of a given number using a while
num = int(input("Enter a number to find its factorial: "))
factorial = 1
current_number = 1
while current_number <= num:
    factorial *= current_number
    current_number += 1
print(f"The factorial of {num} is: {factorial}")

```

```python
#Accept numbers using input() function until the user enters 0. If user input 0 then break the while loop and display the sum of all the numbers.

total_sum = 0
while True:
    num = int(input("Enter a number (enter 0 to stop): "))
    if num == 0:
        break
    total_sum += num

print(f"The sum of all numbers entered is: {total_sum}")
```

```python
# Python program to check if a given number is an Armstrong number

num = int(input("Enter a number to check for Armstrong: "))
order = len(str(num))

sum_of_powers = 0
temp_num = num
while temp_num > 0:
    digit = temp_num % 10
    sum_of_powers += digit ** order
    temp_num //= 10

if num == sum_of_powers:
    print(f"{num} is an Armstrong number.")
else:
    print(f"{num} is not an Armstrong number.")
```

```python
# Python program to get the Fibonacci series between 0 to 50

a, b = 0, 1
print("Fibonacci series (0 to 50):")
while a <= 50:
    print(a, end=" ")
    a, b = b, a + b
print()
```
