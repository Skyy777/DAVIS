Cluster

K-Means Implementation in SciPy

```python
import numpy as np
from numpy import vstack,array
from numpy.random import rand
from scipy.cluster.vq import kmeans,vq,whiten
data1 = vstack((rand(150,2) + array([.5,.5]),rand(150,2)))
data1 = whiten(data1)
centroids,_ = kmeans(data1,2)
result,_ = vq(data1,centroids)
print(result)
```

```python
import numpy as np
time_step = 0.02
period = 5.
time_vec = np.arange(0,20,time_step)
sig = np.sin(2 * np.pi / period * time_vec) + 0.5 * np.random.randn(time_vec.size)
print(sig.size)
```

```python
from scipy import fftpack
sample_freq = fftpack.fftfreq(sig.size,d = time_step)
sig_fft = fftpack.fft(sig)
print(sig_fft)
```

Interpolate

```python
from scipy import interpolate
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,4,12)
y=np.cos(x**2/3+4)
print(x,y)
```

```python
plt.plot(x,y,'o')
plt.show()
```

1D interpolation

```python
f1=interpolate.interp1d(x,y,kind='nearest')
f2=interpolate.interp1d(x,y,kind='linear')
f3=interpolate.interp1d(x,y,kind='cubic')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
fun1= interp1d(x,y,kind='nearest')
fun2= interp1d(x,y,kind='linear')
fun3= interp1d(x,y,kind='cubic')
newx=np.linspace(0,4,50)
plt.plot(x,y,'o',newx,fun1(newx),newx,fun2(newx),newx,fun3(newx))
plt.legend(['data','nearest','linear','cubic'],loc='best')
plt.show()
```

Splines

```python
from scipy.interpolate import UnivariateSpline
x=np.linspace(-3,3,50)
y=np.exp(-x**2)+0.1*np.random.randn(50)
plt.plot(x,y,'ro',ms=5)
plt.show()
```

```python
spl=UnivariateSpline(x,y)
xs=np.linspace(-3,3,100)
plt.plot(xs,spl(xs),'g',lw=5)
plt.show()
```

```python
spl.set_smoothing_factor(0.5)
plt.plot(xs,spl(xs),'g',lw=5)
plt.show()
```
