What is Object Oriented Programming? Explain class and object with a simple Python example.


```python
class person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = person("Sahil", 22)

print(p1.name)
print(p1.age)
```

What is encapsulation? How does Python achieve data hiding? Explain with an example.

```python
class Car:
    def __init__(self, make, _model):
        self.make = make
        self._model = _model

my_car = Car("Toyota", "Supra")
print(f"Make: {my_car.make}, Model: {my_car._model}")
```

What is inheritance? Write a Python program to show single inheritance.

```python
class Parent:
    def greeting(self):
        return "Hello from Parent"

class Child(Parent):
    def say_something(self):
        return "I am the Child"

my_child = Child()
print(my_child.greeting())
print(my_child.say_something())
```

What is an exception in Python? Why is exception handling required?


```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
finally:
    print("Execution complete.")
```

Write a Python program to handle division by zero and invalid input using exception handling.

```python
def safe_division():
    try:
        num1_str = input("Enter the numerator: ")
        num2_str = input("Enter the denominator: ")
        num1 = float(num1_str)
        num2 = float(num2_str)
        result = num1 / num2
        print(f"The result of {num1} / {num2} is: {result}")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero. Please provide a non-zero denominator.")
    except ValueError:
        print("Error: Invalid input. Please enter valid numbers for both numerator and denominator.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("Division attempt concluded.")
safe_division()
```
