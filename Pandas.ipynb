Pandas

```python
import pandas as pd
import numpy as np
```

```python
data=np.array(['g','e','e','k','s','f','o','r','g','e','k','s'])
ser=pd.Series(data)
print(ser)
```

```python
data=np.array(['g','e','e','k','s','f','o','r','g','e','k','s'])
ser=pd.Series(data)
print(ser[:5])
```

```python
data=np.array(['g','e','e','k','s','f','o','r','g','e','k','s'])
ser=pd.Series(data)
print(ser[5:])
```

Binary Operation on Series

```python
data=pd.Series([5,2,3,7],index=['a','b','c','d'])
data1=pd.Series([1,6,4,9],index=['a','b','d','e'])
print(data,data1)
```

```python
data.add(data1,fill_value=0)
```

```python
data.sub(data1,fill_value=0)
```

```python
data.mul(data1,fill_value=0)
```

```python
data.div(data1,fill_value=0)
```

```python
data.pow(data1,fill_value=0)
#exponential power
```

```python
data.mod(data1,fill_value=0)
```
