Binary Operation on Series

```python
data=pd.Series([5,2,3,7],index=['a','b','c','d'])
data1=pd.Series([1,6,4,9],index=['a','b','d','e'])
print(data,data1)
```

```python
data.add(data1,fill_value=0)
```

```python
data.sub(data1,fill_value=0)
```

```python
data.mul(data1,fill_value=0)
```

```python
data.div(data1,fill_value=0)
```

```python
data.pow(data1,fill_value=0)
#exponential power
```

```python
data.mod(data1,fill_value=0)
```

Joins

```python
df1=pd.DataFrame({'lkey':['foo','bar','baz','foo'],'value':[1,2,3,5]})
df2=pd.DataFrame({'rkey':['foo','bar','baz','foo'],'value':[5,6,7,8]})
df1.join(df2,lsuffix='_l',rsuffix='_r')
```

```python
df1=pd.DataFrame({'key':['K0','K1','K5','K3','K4','K2'],
                  'A':['A0','A1','A2','A3','A4','A5']})
df2=pd.DataFrame({'key':['K0','K1','K2'],
                  'B':['B0','B1','B2']})
```

```python
df1.join(df2,lsuffix='_l',rsuffix='_r')
```

```python
df1=pd.DataFrame({'key':['K0','K1','K5','K3','K4','K2'],
                  'A':['A0','A1','A2','A3','A4','A5']},
                 index=[0,1,2,3,4,5])
df2=pd.DataFrame({'key':['K0','K1','K2'],
                  'B':['B0','B1','B2']},
                 index=[0,1,2])
df1.join(df2,lsuffix='_l',rsuffix='_r')
```

```python
#inner join
df1.join(df2,lsuffix='_l',rsuffix='_r',how='inner')
```

```python
#outer join
df1.join(df2,lsuffix='_l',rsuffix='_r',how='outer')
```

```python
#join on column
df1.set_index('key').join(df2.set_index('key'))
```

Concat

```python
m1=pd.Series(['m','n'])
m2=pd.Series(['p','q'])
pd.concat([m1,m2])
```

```python
m1=pd.DataFrame({
    'Name':['Alex','Amy','Allen','Alice','Ayoung'],
    'subject_id':['sub1','sub2','sub4','sub6','sub5'],
    'Marks_scored':[98,90,87,69,78]},
                index=[1,2,3,4,5])
m2=pd.DataFrame({
    'Name':['Billy','Brian','Bran','Bryce','Betty'],
    'subject_id':['sub2','sub4','sub3','sub6','sub5'],
    'Marks_scored':[89,80,79,97,88]},
                index=[1,2,3,4,5])
pd.concat([m1,m2])
```

```python
pd.concat([m1,m2],axis=1)
```

```python
df1=pd.DataFrame({'A':[1,2,3],'B':[1,2,3]})
df2=pd.DataFrame({'A':[4,5,6],'B':[4,5,6]})
pd.concat([df1,df2])
```

```python
df=pd.concat([df1,df2],keys=['x','y'])
df
```

```python
df=pd.concat([df1,df2],ignore_index=True)
df
```
